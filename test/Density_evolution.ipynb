# Density Evolution

```python
import numpy as np
import pandas as pd
from scipy.stats import poisson
```

## DE: Uncoupled & w/o 1st edge

```python
def density_evolution(alpha, k=3, num_iterations=20, delta=0.03):
    """
    Perform the density evolution:
      C_t = 1 - exp(-lambda * S_{t-1}) * (1 - delta)
      S_t = C_t^2
    where lambda = k * alpha.
    
    Returns arrays C, S, conv (where conv = C**k), each of length num_iterations.
    S array corresponds to S_{t-1} for each iteration.
    """
    lmbda = k * alpha
    C = []
    S = [1.0]  
    for _ in range(num_iterations):
        c_next = 1 - np.exp(-lmbda * S[-1]) * (1 - delta)
        C.append(c_next)
        S.append(c_next ** (k-1))

    S = np.array(S[:-1])
    C = np.array(C)
    conv = C ** k
    return C, S, conv
```

```python
def print_res(*arrays):
    """
    Print results in table form. Supports two formats:
      1) Single sequence: print_res(C_vals, S_vals, conv_vals)
         - columns: Iteration, C, S, Conv
      2) Double sequence: print_res(C1_vals, C2_vals, S1_vals, S2_vals, conv_vals)
         - columns: Iteration, C1, C2, S1, S2, Conv
    """
    if len(arrays) == 3:
        C_vals, S_vals, conv_vals = arrays
        df = pd.DataFrame({
            'Iteration': np.arange(1, len(C_vals) + 1),
            'C': np.round(C_vals, 4),
            'S': np.round(S_vals, 4),
            'Conv': np.round(conv_vals, 4)
        })
        print(df.to_string(index=False))

    elif len(arrays) == 5:
        C1_vals, C2_vals, S1_vals, S2_vals, conv_vals = arrays
        df = pd.DataFrame({
            'Iteration': np.arange(1, len(C1_vals) + 1),
            'C1': np.round(C1_vals, 4),
            'C2': np.round(C2_vals, 4),
            'S1': np.round(S1_vals, 4),
            'S2': np.round(S2_vals, 4),
            'Conv': np.round(conv_vals, 4)
        })
        print(df.to_string(index=False))
```

```python
alpha = 0.78
delta = 0.01
print(f"alpha (load): {alpha}, overhead: {1/alpha-1}, delta: {delta}")
C_vals, S_vals, conv_vals = density_evolution(alpha, k=3, num_iterations=30, delta=delta)
print_res(C_vals, S_vals, conv_vals)
```

## DE: uncoupled & w/ 1st edge

```python
def density_evolution_with_listed_edge(alpha, k=3, num_iterations=20, delta=0.03, max_i=11):
    """
    Perform the density evolution:
      λ = (k-1) * alpha
      β_i = i * Poisson(i; λ) / λ,  for i = 0..max_i
      C2_t = 1 - exp(-λ * S2_{t-1}) * (1 - alpha * S1_{t-1}) * (1 - delta)
      C1_t = 1 - exp(-λ * S2_{t-1}) * (1 - delta)
      S1_t = (C2_t)^(k-1)
      S2_t = C1_t * C2_t^(k-2)

    Returns numpy arrays: C1, C2, S1, S2, conv (where conv[t] = C1[t] * C2[t]^(k-1)),
    each of length num_iterations. 
    """
    lmbda = 2 * alpha

    # Precompute beta_i for i = 0..max_i
    beta = np.array([i * poisson.pmf(i, lmbda) / lmbda for i in range(max_i + 1)])

    # Initialize lists
    S1 = [1.0]   
    S2 = [1.0]  
    C1 = []
    C2 = []

    for t in range(1, num_iterations + 1):
        # Compute C2_t
        c2_next = 1 - np.exp(-lmbda * S2[-1]) * (1 - alpha * S1[-1]) * (1 - delta)
        C2.append(c2_next)

        # Compute C1_t
        c1_next = 1 - np.exp(-lmbda * S2[-1]) * (1 - delta)
        C1.append(c1_next)

        # Update S1_t and S2_t
        S1.append(c2_next ** (k-1))
        S2.append(c1_next * c2_next ** (k-2))

    S1_arr = np.array(S1[:-1])   
    S2_arr = np.array(S2[:-1])   
    C1_arr = np.array(C1)
    C2_arr = np.array(C2)
    conv = C1_arr * (C2_arr ** (k-1))

    return C1_arr, C2_arr, S1_arr, S2_arr, conv
```

```python
alpha = 0.8835
delta = 0.0
print(f"alpha (load): {alpha}, overhead: {1/alpha-1}, delta: {delta}")
C1_arr, C2_arr, S1_arr, S2_arr, conv = density_evolution_with_listed_edge(alpha, k=3, delta=delta, num_iterations=240)
print_res(C1_arr, C2_arr, S1_arr, S2_arr, conv)
```
